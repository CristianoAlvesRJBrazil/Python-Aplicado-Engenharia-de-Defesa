## 2.1 Variáveis e tipos de dados

Um programa manipula dados — uma distância, uma velocidade, o nome de um sensor. Para trabalhar com um dado, guardamo-lo com um nome: a **variável**. Cria-se com um nome, o sinal de igual e o valor.

**Listagem 2.1 — Atribuição de variáveis.**

```python
velocidade_no = 18          # velocidade de um navio, em nós
nome_sensor = "Radar-A1"    # identificação de um sensor
alvo_detectado = True       # houve detecção?
latitude = -22.9            # coordenada, em graus

print(velocidade_no, nome_sensor, alvo_detectado, latitude)
```

Não foi preciso declarar o tipo de antemão: o Python descobre sozinho a partir do valor — a **tipagem dinâmica**.

### 2.1.1 Os quatro tipos fundamentais
- **`int`** — inteiros: `18`, `-3`, `0`, `3_600_000`. Contagens, identificadores, quantidades.
- **`float`** — reais: `0.007`, `-22.9`, `1852.0`. Separador decimal é o **ponto**. Medidas físicas.
- **`str`** — texto entre aspas: `"Radar-A1"`, `'Marinha'`. Nomes, rótulos, mensagens.
- **`bool`** — lógicos: `True` / `False`. Base de toda decisão.

A função embutida `type` revela o tipo de qualquer valor.

**Listagem 2.2 — Descobrindo o tipo de um dado.**

```python
print(type(velocidade_no))   # <class 'int'>
print(type(latitude))        # <class 'float'>
print(type(nome_sensor))     # <class 'str'>
print(type(alvo_detectado))  # <class 'bool'>
```

> 📝 **Nota** — O tipo de uma variável pode mudar ao longo do programa (é o valor que carrega o tipo, não o nome). Isso dá flexibilidade, mas exige atenção: atribuir um texto a uma variável que antes guardava número costuma gerar erros sutis. Mantenha cada variável fiel a um único papel.

> ✅ **Boa prática (nomes)** — Dê às variáveis nomes que digam o que significam: `velocidade_no = 18` dispensa explicação, `x = 18` não. Convenção: minúsculas com sublinhado (`distancia_alvo_km`). Em código auditável — como o de defesa — um nome claro é a primeira linha de documentação.

## 2.2 Operadores

### 2.2.1 Aritméticos

| Operador | Operação | Exemplo |
|---|---|---|
| `+` | soma | `10 + 3 → 13` |
| `-` | subtração | `10 - 3 → 7` |
| `*` | multiplicação | `10 * 3 → 30` |
| `/` | divisão (real) | `10 / 3 → 3.333...` |
| `//` | divisão inteira | `10 // 3 → 3` |
| `%` | resto (módulo) | `10 % 3 → 1` |
| `**` | potência | `10 ** 3 → 1000` |

A velocidade naval costuma ser dada em **nós** (1 nó = 1,852 km/h). Convertamos:

**Listagem 2.3 — Conversão de unidades: nós para km/h.**

```python
velocidade_no = 18     # velocidade em nós
fator = 1.852          # 1 nó = 1,852 km/h

velocidade_kmh = velocidade_no * fator
print("Velocidade:", velocidade_kmh, "km/h")
```

> ⚠️ **Armadilha comum** — Cuidado com `/` vs `//`. A divisão `/` **sempre** produz `float`, mesmo exato (`10 / 2 → 5.0`, não `5`). Já `//` descarta a parte fracionária. Trocar uma pela outra é um erro **silencioso**: o programa não falha, só entrega um número ligeiramente errado.

```python
# Veja a diferença na prática:
print(10 / 2)    # 5.0  (float)
print(10 // 3)   # 3    (inteiro, descarta fração)
print(10 % 3)    # 1    (resto)
```

### 2.2.2 Comparação
Produzem sempre um `bool` — matéria-prima das decisões.

| Operador | Significado | Exemplo |
|---|---|---|
| `==` | igual a | `18 == 18 → True` |
| `!=` | diferente de | `18 != 20 → True` |
| `>` `<` | maior / menor | `18 > 20 → False` |
| `>=` `<=` | maior-igual / menor-igual | `18 >= 18 → True` |

> ⚠️ **Armadilha comum** — Não confunda `=` (atribui) com `==` (compara). `if velocidade = 18` é erro de sintaxe; o correto é `if velocidade == 18`.

```python
print(18 == 18)   # True
print(18 != 20)   # True
print(18 > 20)    # False
print(18 <= 17)   # False
```

### 2.2.3 Lógicos
Para combinar condições, o Python usa **palavras**: `and` (e), `or` (ou), `not` (não).

**Listagem 2.4 — Combinando condições com operadores lógicos.**

```python
velocidade_kmh = 33.3
em_area_restrita = True

# Verdadeira só se ambas as partes forem verdadeiras
alerta = (velocidade_kmh > 30) and em_area_restrita
print("Disparar alerta?", alerta)   # True
```

## 2.3 Entrada e saída de dados

### 2.3.1 Lendo com `input`
`input` pausa o programa, exibe uma mensagem e devolve o que for digitado **sempre como `str`**.

**Listagem 2.5 — Lendo um valor do usuário.** *(ao executar, digite um número e tecle Enter)*

```python
texto = input("Informe a velocidade em nós: ")
velocidade_no = float(texto)        # converte o texto em número real
velocidade_kmh = velocidade_no * 1.852
print("Equivale a", velocidade_kmh, "km/h")
```

A conversão é essencial: `input` entrega texto. As funções `int`, `float` e `str` convertem (*cast*) valores entre tipos.

> ⚠️ **Armadilha comum** — Tudo o que vem de `input` é texto. `"2" + "3"` resulta em `"23"`, não `5`. Sempre converta a entrada para o tipo esperado, e o quanto antes.

```python
# Demonstração da armadilha:
print("2" + "3")          # "23"  (concatenação de texto)
print(int("2") + int("3")) # 5    (soma de números)
```

### 2.3.2 Saída formatada com f-strings
As **f-strings** (prefixadas por `f`) avaliam qualquer expressão entre chaves e a inserem no texto.

**Listagem 2.6 — Saída formatada com f-strings.**

```python
velocidade_no = 18
velocidade_kmh = velocidade_no * 1.852

print(f"O navio a {velocidade_no} nós navega a {velocidade_kmh:.1f} km/h.")
```

O trecho `:.1f` é um **especificador de formato**: exibe o número como real com **uma casa decimal**. As f-strings serão usadas ao longo de todo o livro.

## 2.4 Estruturas condicionais

Decidir é da essência de qualquer sistema de apoio à decisão. Use `if` + condição + dois-pontos; o bloco dependente vem **indentado**.

**Listagem 2.7 — Decisão simples com if/else.**

```python
velocidade_kmh = 42.0
limite_kmh = 40.0

if velocidade_kmh > limite_kmh:
    print("ALERTA: velocidade acima do limite.")
else:
    print("Velocidade dentro do limite.")
```

Com mais de duas possibilidades, encadeiam-se condições com `elif` (contração de *else if*).

**Listagem 2.8 — Classificação em faixas com elif.**

```python
velocidade_kmh = 75.0

if velocidade_kmh < 30:
    classe = "lento"
elif velocidade_kmh < 60:
    classe = "moderado"
elif velocidade_kmh < 100:
    classe = "rápido"
else:
    classe = "muito rápido"

print(f"Contato classificado como: {classe}.")
```

As condições são avaliadas de cima para baixo; assim que uma é `True`, seu bloco roda e as demais são ignoradas. O `else` final é opcional ("nenhuma das anteriores").

> 🛡️ **Contexto de defesa** — A lógica condicional é o coração de um sistema de apoio à decisão. Regras como *"se o contato está em área restrita E acima da velocidade permitida, então elevar o alerta"* viram `if`/`elif`/`else`. Em um sistema auditável, a clareza dessas regras é o que permite **explicar depois por que** uma decisão foi recomendada.

> ⚠️ **Armadilha comum** — Em Python, **a indentação é sintaxe**, não decoração. O bloco de um `if` é definido pelo recuo (em geral, 4 espaços). Misturar espaços e tabs, ou alinhar de forma inconsistente, provoca `IndentationError`.

## 2.5 Estruturas de repetição

### 2.5.1 O laço `for`
Percorre os elementos de uma coleção, um a um.

**Listagem 2.9 — Percorrendo leituras de telemetria com for.**

```python
# Leituras de velocidade (km/h) recebidas de um sensor
leituras = [28.4, 33.1, 41.7, 39.0, 45.2]
limite = 40.0

for leitura in leituras:
    if leitura > limite:
        print(f"Leitura {leitura} km/h: ACIMA do limite.")
    else:
        print(f"Leitura {leitura} km/h: normal.")
```

Para repetir um número fixo de vezes, usa-se `range`, que gera inteiros.

**Listagem 2.10 — Repetição contada com range.**

```python
# Tabela de conversão: 1 a 5 nós em km/h
for nos in range(1, 6):     # gera 1, 2, 3, 4, 5
    kmh = nos * 1.852
    print(f"{nos} nó(s) = {kmh:.3f} km/h")
```

> 📝 **Nota** — `range(1, 6)` gera de 1 a **5** (o limite final é **exclusivo**). Já `range(5)` gera de **0 a 4**.

### 2.5.2 O laço `while`
Repete **enquanto** uma condição for verdadeira — ideal quando não se sabe o número de repetições de antemão.

**Listagem 2.11 — Repetição condicional com while.**

```python
contador = 0
acumulado = 0.0

# Soma leituras simuladas até ultrapassar 100 km/h acumulados
while acumulado < 100:
    acumulado = acumulado + 25.0   # cada leitura adiciona 25 km/h
    contador = contador + 1

print(f"Foram necessárias {contador} leituras para ultrapassar 100.")
```

> ⚠️ **Armadilha comum** — Todo `while` precisa, em algum momento, tornar a condição **falsa** — senão entra em **laço infinito**. Aqui, é a linha que incrementa `acumulado` que garante o fim. (No Colab, um laço infinito exige interromper a execução manualmente: ⏹ *Parar*.)

## 2.6 Funções: organizando e reutilizando código

Uma **função** é um trecho de código com nome, que recebe **parâmetros**, faz algo e em geral **devolve** um resultado. Evitam repetição e dão nome às intenções.

**Listagem 2.12 — Definindo uma função de conversão.**

```python
def nos_para_kmh(velocidade_no):
    """Converte uma velocidade de nós para km/h."""
    return velocidade_no * 1.852

# Agora a conversão é uma chamada, reutilizável quantas vezes quisermos.
# Arredondamos na exibição com round(_, 3) para evitar o "ruído" do ponto
# flutuante: sem isso, nos_para_kmh(25) apareceria como 46.300000000000004.
print(round(nos_para_kmh(18), 3))   # 33.336
print(round(nos_para_kmh(25), 3))   # 46.3
print(round(nos_para_kmh(10), 3))   # 18.52
```

`def` inicia a definição; seguem nome, parâmetros e dois-pontos. O bloco indentado é o corpo. A linha entre aspas triplas é a **docstring** (documentação). `return` indica o valor devolvido.

Funções podem ter **vários parâmetros** e **valores padrão**. Exemplo: alcance balístico simplificado, $R = \dfrac{v^2 \sin(2\theta)}{g}$.

**Listagem 2.13 — Alcance balístico simplificado.**

```python
import math   # biblioteca padrão de funções matemáticas

def alcance(velocidade_ms, angulo_graus, g=9.81):
    """Alcance horizontal (m) de um projétil, sem resistência do ar."""
    angulo_rad = math.radians(angulo_graus)
    return (velocidade_ms ** 2) * math.sin(2 * angulo_rad) / g

# Velocidade de 50 m/s, ângulo de 45 graus
print(f"Alcance: {alcance(50, 45):.1f} m")
```

`import math` torna disponível a biblioteca matemática (`radians`, `sin`) — o Python traz "baterias inclusas". O parâmetro `g=9.81` tem **valor padrão**: se quem chama não o informa, adota-se 9,81 m/s²; mas poderia passar outro valor (p. ex., outro corpo celeste).

> ✅ **Boa prática** — Uma função deve fazer **uma** coisa e fazê-la bem; o nome deve dizer qual. Se você explica o que ela faz usando "e" ("converte **e** imprime **e** grava"), provavelmente são três funções. Funções pequenas e bem-nomeadas são mais fáceis de testar, reaproveitar e **auditar**.

## 2.7 Juntando tudo: um pequeno processador de telemetria

Vamos combinar variáveis, operadores, uma função, um laço e uma condicional em um programa que processa leituras e produz um relatório — um embrião da lógica do miniprojeto.

**Listagem 2.14 — Processador de telemetria reunindo os fundamentos.**

*(Reutiliza `nos_para_kmh`, definida na Listagem 2.12 — execute aquela célula antes desta.)*

```python
def classificar(velocidade_kmh, limite):
    """Devolve 'alerta' se acima do limite, senão 'normal'."""
    if velocidade_kmh > limite:
        return "alerta"
    return "normal"

leituras_no = [12, 18, 24, 9, 27]   # leituras em nós
limite_kmh = 40.0
total_alertas = 0

print("Relatório de telemetria")
print("-" * 30)

for leitura_no in leituras_no:
    kmh = nos_para_kmh(leitura_no)          # reutiliza a função da Seção anterior
    estado = classificar(kmh, limite_kmh)
    if estado == "alerta":
        total_alertas = total_alertas + 1
    print(f"{leitura_no:>3} nós -> {kmh:6.1f} km/h [{estado}]")

print("-" * 30)
print(f"Total de leituras em alerta: {total_alertas}")
```

Em pouco mais de vinte linhas, o programa converte unidades, classifica cada leitura, conta alertas e apresenta o resultado de forma legível. Note os recursos de f-strings: `:>3` alinha o número à direita em 3 colunas; `:6.1f` reserva 6 colunas e 1 casa decimal — detalhes que transformam saída desalinhada em **relatório apresentável**. É essa combinação de elementos simples que, ampliada, dará origem ao sistema de apoio à decisão do miniprojeto.
