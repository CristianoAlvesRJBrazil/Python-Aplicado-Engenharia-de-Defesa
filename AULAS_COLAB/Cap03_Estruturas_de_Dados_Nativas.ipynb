## 3.1 Listas: a estrutura de trabalho

A **lista** guarda uma sequência **ordenada** de elementos (de qualquer tipo) e é **mutável**. Escrita entre **colchetes**, com vírgulas.

**Listagem 3.1 — Criando listas.**

```python
# Um inventário de meios (embarcações de uma flotilha)
meios = ["Fragata", "Corveta", "Navio-Patrulha", "Rebocador"]

# Leituras de velocidade de um sensor (km/h)
leituras = [28.4, 33.1, 41.7, 39.0, 45.2]

# Uma lista pode até misturar tipos, embora raramente convenha
registro = ["Radar-A1", 3, True]

print(meios)
print("Quantidade de meios:", len(meios))
```

`len` devolve o número de elementos.

### 3.1.1 Acessando por índice
O **índice** começa em **zero**. Índices negativos contam do fim (`-1` é o último).

**Listagem 3.2 — Acesso por índice.**

```python
meios = ["Fragata", "Corveta", "Navio-Patrulha", "Rebocador"]

print(meios[0])    # Fragata (primeiro)
print(meios[2])    # Navio-Patrulha
print(meios[-1])   # Rebocador (último)
print(meios[-2])   # Navio-Patrulha (penúltimo)
```

> ⚠️ **Armadilha comum** — A contagem começa em **zero**. Em uma lista de 4 elementos, índices válidos vão de 0 a 3 (e de −1 a −4). `meios[4]` provoca `IndexError: list index out of range` — o clássico "erro de um a mais". Desconfie sempre dos limites.

### 3.1.2 Fatiamento (slice)
Notação `[início:fim]` — o índice **final é exclusivo** (como no `range`).

**Listagem 3.3 — Fatiamento de listas.**

```python
leituras = [28.4, 33.1, 41.7, 39.0, 45.2]

print(leituras[0:3])   # [28.4, 33.1, 41.7] (índices 0,1,2)
print(leituras[2:])    # [41.7, 39.0, 45.2] (do índice 2 ao fim)
print(leituras[:2])    # [28.4, 33.1] (do início ao índice 1)
print(leituras[-2:])   # [39.0, 45.2] (as duas últimas)
```

Útil para **janelas de dados**: as últimas N leituras, um intervalo temporal específico etc.

### 3.1.3 Modificando listas
Por serem mutáveis, têm métodos (notação de ponto):

| Método | O que faz |
|---|---|
| `lista.append(x)` | adiciona `x` ao final |
| `lista.insert(i, x)` | insere `x` na posição `i` |
| `lista.remove(x)` | remove a 1ª ocorrência de `x` |
| `lista.pop(i)` | remove e **devolve** o elemento de `i` |
| `lista.sort()` | ordena no lugar |
| `lista.reverse()` | inverte a ordem |
| `lista.count(x)` | conta ocorrências de `x` |

**Listagem 3.4 — Modificando uma lista de ocorrências.**

```python
ocorrencias = ["contato-norte", "contato-leste"]

ocorrencias.append("contato-sul")          # adiciona ao final
ocorrencias.insert(0, "contato-urgente")   # insere no início
print(ocorrencias)

ocorrencias.remove("contato-leste")        # remove pelo valor
ultima = ocorrencias.pop()                 # remove e devolve a última
print("Removida:", ultima)
print("Restantes:", ocorrencias)
```

> ⚠️ **Armadilha comum** — `sort` e `reverse` alteram a lista **no lugar** e devolvem `None`. `leituras = leituras.sort()` é erro: a variável vira `None` e os dados se perdem. Use apenas `leituras.sort()`; para uma cópia ordenada, `sorted(leituras)`.

### 3.1.4 Percorrendo com índice
Quando precisar de índice **e** valor, use `enumerate`:

**Listagem 3.5 — enumerate.**

```python
leituras = [28.4, 33.1, 41.7, 39.0, 45.2]

for indice, valor in enumerate(leituras):
    print(f"Leitura {indice}: {valor} km/h")
```

## 3.2 Compreensão de listas

Construir uma nova lista a partir de outra — convertendo ou filtrando — é tarefa comum. A **compreensão de lista** faz isso de forma concisa.

**Listagem 3.6 — Do laço à compreensão.**

```python
leituras_no = [12, 18, 24, 9, 27]

# Forma tradicional, com laço
kmh = []
for no in leituras_no:
    kmh.append(no * 1.852)

# Forma idiomática, com compreensão de lista
kmh = [no * 1.852 for no in leituras_no]

print(kmh)
```

A compreensão diz, em uma linha: *"para cada `no` em `leituras_no`, calcule `no * 1.852`"*. Também aceita **filtro** com `if`.

**Listagem 3.7 — Compreensão com filtro.**

```python
leituras = [28.4, 33.1, 41.7, 39.0, 45.2]

# Apenas as leituras acima de 40 km/h
acima = [x for x in leituras if x > 40]
print(acima)        # [41.7, 45.2]
print(len(acima))   # 2
```

> ✅ **Boa prática** — A compreensão é mais legível que o laço **quando a lógica é simples** (uma transformação, um filtro, ou ambos). Se a regra exige vários `if`, cálculos intermediários ou aninhamentos profundos, volte ao laço: clareza vale mais que concisão. A compreensão é ferramenta de elegância, não teste de esperteza.

## 3.3 Tuplas: dados que não mudam

A **tupla** é como uma lista escrita com **parênteses**, mas **imutável**: depois de criada, não se altera, adiciona ou remove elementos.

**Listagem 3.8 — Criando e usando tuplas.**

```python
# Uma coordenada geográfica: latitude e longitude
posicao = (-22.9, -43.2)

print(posicao[0])   # -22.9 (acesso por índice, como na lista)
print(posicao[1])   # -43.2

# Tentar alterar provoca erro:
# posicao[0] = -23.0  -> TypeError: 'tuple' object does not support item assignment
```

Por que uma estrutura imutável? Porque há dados que **não devem mudar**: uma coordenada, um par de constantes, um registro que precisa permanecer íntegro. Em sistemas auditáveis, **poder afirmar que um dado não foi alterado** tem valor concreto.

### 3.3.1 Desempacotamento
Atribuir, de uma vez, cada elemento a uma variável.

**Listagem 3.9 — Desempacotamento.**

```python
posicao = (-22.9, -43.2)

latitude, longitude = posicao   # desempacota nos dois nomes
print(f"Lat {latitude}, Lon {longitude}")

# Útil também para trocar valores sem variável auxiliar
a, b = 10, 20
a, b = b, a
print(a, b)   # 20 10
```

> 📝 **Nota** — Funções que devolvem mais de um valor o fazem retornando uma **tupla**. Ex.: `return menor, maior`; quem chama recebe ambos com `m, M = estatisticas(leituras)`. Padrão frequente.

## 3.4 Dicionários: dados com rótulo

Listas e tuplas guardam dados **por posição**. O **dicionário** guarda **pares chave–valor** — acesso por um **nome significativo**. Escrito entre **chaves**.

**Listagem 3.10 — Um dicionário descrevendo uma ocorrência.**

```python
ocorrencia = {
    "id": 101,
    "sensor": "Radar-A1",
    "tipo": "contato de superfície",
    "velocidade_kmh": 44.4,
    "em_area_restrita": True,
}

print(ocorrencia["sensor"])          # Radar-A1
print(ocorrencia["velocidade_kmh"])  # 44.4
```

Compare `ocorrencia["sensor"]` com um hipotético `ocorrencia[1]`: o dicionário torna o código **autoexplicativo**. Chaves costumam ser textos; valores podem ser de qualquer tipo (inclusive listas ou outros dicionários).

### 3.4.1 Acessando, adicionando e alterando
**Listagem 3.11 — Manipulando um dicionário.**

```python
ocorrencia = {"id": 101, "sensor": "Radar-A1", "velocidade_kmh": 44.4}

# Adicionar ou alterar é igual: atribuir a uma chave
ocorrencia["classe"] = "rápido"          # nova chave
ocorrencia["velocidade_kmh"] = 46.0      # altera valor existente

# Acesso seguro com .get (evita erro se a chave não existir)
print(ocorrencia.get("latitude", "não informada"))

# Remover uma chave
del ocorrencia["id"]
print(ocorrencia)
```

> ⚠️ **Armadilha comum** — Acessar chave inexistente com colchetes (`ocorrencia["latitude"]`) provoca `KeyError`. Sem certeza de que a chave existe, prefira `.get(chave, padrao)`. Em dados do mundo real (sensores, arquivos, digitação), **nunca presuma** que um campo está presente.

### 3.4.2 Percorrendo dicionários
`.items()` entrega, a cada volta, a chave e o valor.

**Listagem 3.12 — Percorrendo com items().**

```python
ocorrencia = {
    "sensor": "Radar-A1",
    "tipo": "contato de superfície",
    "velocidade_kmh": 44.4,
}

for chave, valor in ocorrencia.items():
    print(f"{chave}: {valor}")
```

`.keys()` e `.values()` entregam apenas chaves ou apenas valores.

### 3.4.3 O padrão **lista de dicionários**
A combinação mais importante do capítulo e a base do miniprojeto: cada dicionário é um **registro** (campos rotulados); a lista reúne todos — uma "tabela" de dados em Python.

**Listagem 3.13 — Lista de dicionários: registro de ocorrências.**

```python
ocorrencias = [
    {"id": 1, "sensor": "Radar-A1", "velocidade_kmh": 22.2, "alerta": False},
    {"id": 2, "sensor": "Radar-A1", "velocidade_kmh": 44.4, "alerta": True},
    {"id": 3, "sensor": "Radar-B2", "velocidade_kmh": 50.0, "alerta": True},
]

# Quantas ocorrências estão em alerta?
em_alerta = [o for o in ocorrencias if o["alerta"]]
print(f"Ocorrências em alerta: {len(em_alerta)}")

# Listar os sensores envolvidos em alertas
for o in em_alerta:
    print(f" - {o['sensor']} a {o['velocidade_kmh']} km/h")
```

Compreensão de lista, dicionário e f-string se combinam com naturalidade. Esse trecho já é, em essência, uma **consulta a uma base de dados em memória**.

## 3.5 Conjuntos: unicidade e pertencimento

O **conjunto** (`set`) guarda elementos **únicos** e **não ordenados**. Especialidade: responder com eficiência a *"este elemento já está aqui?"* e *"quais são os distintos?"*.

**Listagem 3.14 — Removendo duplicatas com conjuntos.**

```python
# Sensores que registraram ocorrências (com repetições)
sensores = ["Radar-A1", "Radar-B2", "Radar-A1", "Radar-A1", "Radar-B2"]

# Quais sensores distintos atuaram?
distintos = set(sensores)
print(distintos)                       # {'Radar-A1', 'Radar-B2'}
print("Sensores distintos:", len(distintos))   # 2

# Pertencimento é imediato e eficiente
print("Radar-A1" in distintos)   # True
print("Radar-C3" in distintos)   # False
```

Conjuntos suportam as operações clássicas: **união** (`|`), **interseção** (`&`) e **diferença** (`-`).

**Listagem 3.15 — Operações de conjunto.**

```python
sensores_manha = {"Radar-A1", "Radar-B2", "Sonar-1"}
sensores_tarde = {"Radar-B2", "Sonar-1", "Radar-C3"}

print(sensores_manha & sensores_tarde)   # ativos nos dois turnos
print(sensores_manha | sensores_tarde)   # ativos em algum turno
print(sensores_manha - sensores_tarde)   # só de manhã
```

## 3.6 Escolhendo a estrutura certa

As quatro estruturas não competem — **complementam-se**. Saber escolher é habilidade de engenharia.

| Estrutura | Quando usar | Exemplo de defesa |
|---|---|---|
| **Lista** | sequência ordenada que pode mudar | leituras de um sensor, em ordem de chegada |
| **Tupla** | dados fixos que não devem mudar | uma coordenada (lat, lon); par de constantes |
| **Dicionário** | dados acessados por um rótulo | uma ocorrência com seus campos nomeados |
| **Conjunto** | itens únicos; teste de pertencimento | sensores distintos que atuaram numa missão |

> 🛡️ **Contexto de defesa** — A escolha da estrutura é uma **decisão de modelagem**. Uma frota é uma **lista** de meios; cada meio é um **dicionário** de atributos; sua posição é uma **tupla** imutável; e os tipos de meio presentes formam um **conjunto**. Um modelo que espelha a realidade torna o resto do sistema quase óbvio de escrever; um modelo torto contamina tudo o que vem depois.

## 3.7 Miniprojeto: modelando os dados (Módulo 1)

Damos início ao **miniprojeto integrador**: um sistema de **registro e análise de ocorrências de monitoramento** (ex.: detecções em área de vigilância). *Você é encorajado a adaptá-lo, desde já, ao problema da sua realidade profissional.*

A tarefa do Módulo 1 é **modelar os dados em memória**. A escolha é natural: cada ocorrência é um **dicionário**; o conjunto de ocorrências, uma **lista**. Escreveremos também as primeiras operações.

**Listagem 3.16 — Miniprojeto, Módulo 1: modelo de dados e primeiras operações.**

```python
# ---- Modelo de dados: uma lista de ocorrências (dicionários) ----
ocorrencias = [
    {"id": 1, "sensor": "Radar-A1", "tipo": "superfície",
     "velocidade_kmh": 22.2, "posicao": (-22.9, -43.2)},
    {"id": 2, "sensor": "Radar-A1", "tipo": "superfície",
     "velocidade_kmh": 44.4, "posicao": (-23.1, -43.5)},
    {"id": 3, "sensor": "Radar-B2", "tipo": "aéreo",
     "velocidade_kmh": 50.0, "posicao": (-22.7, -43.0)},
]

LIMITE_KMH = 40.0

def registrar(lista, sensor, tipo, velocidade, posicao):
    """Cria uma nova ocorrência e a adiciona à lista. Devolve o id."""
    novo_id = len(lista) + 1
    lista.append({
        "id": novo_id, "sensor": sensor, "tipo": tipo,
        "velocidade_kmh": velocidade, "posicao": posicao,
    })
    return novo_id

def listar_alertas(lista, limite):
    """Devolve as ocorrências com velocidade acima do limite."""
    return [o for o in lista if o["velocidade_kmh"] > limite]

def sensores_distintos(lista):
    """Devolve o conjunto de sensores presentes nas ocorrências."""
    return {o["sensor"] for o in lista}

# ---- Uso ----
registrar(ocorrencias, "Sonar-1", "submarino", 18.5, (-23.0, -43.3))

print(f"Total de ocorrências: {len(ocorrencias)}")
print(f"Sensores envolvidos: {sensores_distintos(ocorrencias)}")

print("Ocorrências em alerta:")
for o in listar_alertas(ocorrencias, LIMITE_KMH):
    print(f" #{o['id']} {o['sensor']} -> {o['velocidade_kmh']} km/h")
```

Em poucas dezenas de linhas, temos um sistema funcional: modela ocorrências, registra novas, lista as que estão em alerta e identifica os sensores envolvidos. **Cada uma das quatro estruturas comparece** — lista (o conjunto de ocorrências), dicionário (cada ocorrência), tupla (a posição) e conjunto (os sensores distintos). E cada operação é uma função pequena e bem-nomeada, como no Capítulo 2.

> 📝 **Nota** — Este código tem uma fragilidade evidente: **tudo vive apenas na memória**. Ao encerrar o programa, todas as ocorrências se perdem. No próximo capítulo daremos ao sistema a capacidade de **persistir** seus dados — em arquivo e em banco de dados SQLite — para que sobrevivam entre execuções. É o tema do **Módulo 2**.

## 3.8 Resumo do capítulo
- **Lista** `[ ]`: sequência ordenada e mutável; índice a partir de 0; fatiável; métodos `append`, `insert`, `remove`, `pop`, `sort`.
- **Compreensão** `[f(x) for x in lista if cond]`: constrói e filtra em uma linha legível.
- **Tupla** `( )`: sequência imutável — ideal para dados que não devem mudar; suporta desempacotamento.
- **Dicionário** `{chave: valor}`: acesso por rótulo; `.get` para acesso seguro; `.items()` para percorrer.
- **Conjunto** `set`: itens únicos; pertencimento e operações de conjunto eficientes.
- **Lista de dicionários** representa uma "tabela" — base do miniprojeto (Módulo 1).

## Armadilhas comuns
- Esquecer que o índice começa em **zero** (`IndexError`).
- Atribuir o resultado de `sort()` (devolve `None`) — use `sorted()` para cópia.
- Acessar chave inexistente com colchetes (`KeyError`) — prefira `.get`.
- Tentar alterar uma **tupla** (imutável).
- Confundir o limite do fatiamento — em `lista[2:5]`, o índice 5 é **exclusivo**.
- Esperar **ordem** em um conjunto — se a ordem importa, use lista.
